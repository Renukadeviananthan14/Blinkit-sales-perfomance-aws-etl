```python
s3 = boto3.client("s3")
print ("s3 connect successfully")
```

```python
bucket = "blinkit-sales-performance"
file_key = "raw/blinkit_delivery_performance.csv"
print ("connected successfully")
```

```python
raw_data["on_time_rate"] = (
    raw_data[raw_data["delivery_status"] == "On Time"]
    .groupby("delivery_partner_id")
    .size()
    /
    raw_data.groupby("delivery_partner_id").size()
) * 100

on_time_rate.sort_values()
```

```python
 raw_data["on_time_rate"] =  raw_data["on_time_rate"].fillna("Not Bad")
```

```python
raw_data["on_time_rate"].dtype
```

```python
raw_data["on_time_rate"] = pd.to_numeric(
    raw_data["on_time_rate"],
    errors="coerce"
)
```

```python
raw_data=raw_data.sort_values(by="on_time_rate",ascending = True)
```

```python
raw_data ["on_time_rate"] = raw_data["on_time_rate"].fillna(0)
```
